As a reminder, below is a summary of the Nystrom method.

Solve the integral equation 
$$ u(t) - \int_a^b K(t, s)u(s)\mathrm{d}s = f(t) \quad \text{for} \quad t \in [a, b],$$
i.e.
$$ (I - K)u = f.$$

Approximate function $u(t)$ by another function $u_n(t)$ which obeys
$$ u_n(t) - \sum_{j=1}^n w_j K(t, s_j) u_n(s_j) = f(t), $$
i.e.
$$ (I - K_n)u_n = f $$
where the integral has been replaced with a quadrature formula, and the kernel with a rank-$n$ operator.

Then the values of $u_n(t)$ at the nodes $s_i$, $u_i^{(n)} := u_n(s_i), \; i = 1, 2, \ldots n$, satisfy the linear system
$$ \forall i = 1, \ldots, n \quad u_i^{(n)} - \sum_{j=1}^{n} w_j K(s_i, s_j) u_j^{(n)} = f(s_i), $$
i.e.
$$ (I - A)\vec{u}^{(n)} = \vec{f}, $$
with $A_{ij} = K(s_i, s_j) w_j$. Then if any vector $\{ u_i\}_{i=1}^{n}$ solves the above system, then $u_n(t)$ at any $t$ may be reconstructed as
$$u_n(t) = f(t) + \sum_{j=1}^{n} w_j K(t, s_i) u_j^{(n)}.$$
